# Square brackets vs get()
--> Pro tip: Jab unsure ho ke key exist karti hai ya nahi, get() useful hota hai.

```python
student = {"name":"Arham","age":22}
print(student.get("name"))
```

# Dictionary mein data add aur update karna
--> Dictionary mutable hoti hai, yani ismein values add, update aur delete kar sakte hain.
-->Existing value update karna

```python
student = {
    "name":"Arham",
    "city":"Karachi",
    "Phone":"0320202948"
}
student["Phone"] = "03202024089"
print(student)
```

# update() method
Ek saath multiple values add ya update kar sakte hain.

```python
student = {"name": "Arham", "age": 23}
student.update({
    "age":21,
    "name":"Arham Arain",
    "city":"karachi"
})
print(student)
```

# Dictionary se data delete karna
Python mein delete karne ke multiple tareeqe hain.

```python
#pop(key) — Specific key remove
student = {"name": "Arham", "age": 23}
student.pop("age")
print(student)
#remove in the 
```

```python
student = {"name": "Ali", "age": 21}

result = student.popitem()
print(result)
print(student)
```

# del — Specific key delete
# clear() — Dictionary empty

```python
student = {"name": "Ali", "age": 21}

del student["age"]
print(student)

student = {"name": "Ali", "age": 21}

student.clear()
print(student)
```

```python
student = {"name": "Ali", "age": 21, "city": "Karachi"}
print(student.keys())
print(student.values())
print(student.items())


```

```python
student = {"name": "Ali"}

student.setdefault("age", 21)
student.setdefault("name", "Ahmed")

print(student)
```

```python
keys = ["name", "age", "city"]
student = dict.fromkeys(keys,"Unknown")
print(student)
```

# loop in Dict

```python
student = {"name": "Ali", "age": 21, "city": "Karachi"}
for index,key in enumerate (student):
    print({index,key})
```

```python
student = {"name": "Ali", "age": 21, "city": "Karachi"}
for values in student.keys():
    print(values)
```

```python
student = {
    "name": "Arham",
    "age": 23,
    "city": "Karachi"
}

print(len(student))
```

# Nested Dict

```python
student= {
    "student1":{
    "name":"Arham",
    "ID":"CSC-24S-036",
    "Dept":"BSCS",
    "Marks":80
},
"student2": {
    "name":"Hayat ALi",
     "ID":"CSC-24S-002",
    "Dept":"BSCS",
    "Marks":90
}
}

print(student["student2"]["Marks"])
```

```python
students = {
    "student1": {
        "name": "Ali",
        "marks": 80
    },
    "student2": {
        "name": "Sara",
        "marks": 90
    }
}


for student_id, details in students.items():
    print("ID:", student_id)
    print("Name:", details["name"])
    print("Marks:", details["marks"])
```
