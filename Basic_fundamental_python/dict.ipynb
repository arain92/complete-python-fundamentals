# Dictionary kya hoti hai?
Dictionary Python mein ek collection hai jo data ko key-value pairs mein store karti hai.
Real-life example: Student ka biodata.
name
Arham
age
23
degree
BSCS
->Left side = Key, right side = Value.

```python
student = {
    "name" :"Arham",
    "age":23,
    "Degree":"BSCS"
}
print(student)
```

dictionary = {key: value}
- Dictionary ke liye curly brackets {} use hote hain.
- Key aur value ke darmiyan colon : hota hai.
- Har pair ke darmiyan comma , hota hai.
Important: {} empty dictionary hai, empty set nahi.
