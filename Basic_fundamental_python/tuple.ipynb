# Tuple kya hota hai?
--> Python mein Tuple ek collection hai jisme hum multiple values ko ek hi variable mein store kar sakte hain.
Is tuple mein:
- "Arham" = name
- 23 = age
- 85.5 = marks ya score
- "BSCS" = degree
Tuple ko banane ke liye normally round brackets () use karte hain.

```python
student = ("Arham",22,85.6,"BSCS")
print(student[0])  # Output: Arham
```

# List aur Tuple mein difference
-->List — Change ho sakti hai This is aslo Mutable

```python
fruits = ["Apple", "Mango", "Banana"]
fruits[0] = "Orange"

print(fruits)
# ['Orange', 'Mango', 'Banana']
```

-->Tuple — Elements change nahi hote This is aslo called Imutable

```python
fruits = ("Apple", "Mango", "Banana")
fruits[0] = "Orange"  # This will raise a TypeError
print(fruits)
```

# Tuple Indexing like list 

```python
numbers = (10, 20, 30, 40, 50)
print(numbers[0])
print(numbers[1])
print(numbers[2])
print(numbers[3])
```

# Negative Indexing
Negative indexing mein end se values access karte hain.

```python
numbers = (10, 20, 30, 40, 50)

print(numbers[-1])  # 50
print(numbers[-2])  # 40
print(numbers[-3])  # 30
```

# Tuple Slicing
Slicing se hum tuple ka ek portion nikaal sakte hain.
[start : end : step]
start: kahan se shuru karna hai.
end: kahan rukna hai, yeh index include nahi hota.
step: kitne steps aage jaana hai.


```python
numbers = (10, 20, 30, 40, 50, 60)

print(numbers[1:4])
```

# Tuple ke Methods

```python
#counting the number of elements in a tuple
numbers = (10, 20, 30, 40, 50)
print(len(numbers))  # Output: 5
#count the number of occurrences of a specific element in a tuple
numbers = (10, 20, 30, 40, 50, 10, 20, 10)
print(numbers.count(10))  # Output: 3
#index of a specific element in a tuple
numbers = (10, 20, 30, 40, 50)
print(numbers.index(20))  # Output: 1


```

# Kya Tuple mein append(), pop() aur remove() hote hain?
Nahi. Tuples immutable hain, isliye ye methods tuple par available nahi hote.
Note: 

```python
nums = (10, 20, 30)

nums.append(40)  # AttributeError
nums.pop()       # AttributeError
nums.remove(20)  # AttributeError
#Agar tuple ki values change karni hon,
#to ek tareeqa hai usko list mein convert karna.

```

```python
num = (1,2,3,4,5)
print(num)
num = list(num)  # Convert tuple to list
num.insert(0,0)
num.append(6)
print(num)  # This will raise a TypeErro
```

# Tuple Operations
->Concatenation — Do tuples jorna

```python
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
combined_tuple = tuple1 + tuple2
print(combined_tuple)  # Output: (1, 2, 3, 4, 5, 6)
#Repetition — Values repeat karna
nums = (1,2)
print(nums * 3)
```

# Membership — Value exist karti hai ya nahi?

```python
fruits = ("Apple", "Mango", "Banana")
print("Apple" in fruits)  # Output: True
```

# Length — Total elements

```python
nums = [1, 2, 3, 4, 5, 6]
print(len(nums))  # Output: 6
```

# Minimum, maximum aur sum

```python
nums = [1, 2, 3, 4, 5, 6]
print(max(nums))  # Output: 6
print(min(nums))  # Output: 1
print(sum(nums))  # Output: 21
```

# Tuple Packing aur Unpacking
-->packing:Multiple values ko ek tuple mein rakhna.
-->unpacking:Tuple ki values ko separate variables mein nikalna.

```python
#packing
student = "Arham", 23, "BSCS"
print(student)
#unpacking
student = ("Arham", 23, "BSCS")
name, age, degree = student
print(name)
print(age)
print(degree)
```

# Advanced unpacking: *

```python
nums = [1, 2, 3, 4, 5, 6]
first, *middle, last = nums
print(first)   # Output: 1
print(middle)  # Output: [2, 3, 4, 5]
print(last)    # Output: 6
```

# Tuple ke saath Loops


```python
fruits = ("Apple", "Mango", "Banana")
for fruit in fruits:
    print(fruit)
```

# Index aur value ek saath: enumerate()

```python
fruits = ("Apple", "Mango", "Banana")
for index, fruit in enumerate(fruits):
    print(index, fruit)
```

# Tuple par condition

```python
nums = [1, 2, 3, 4, 5, 6]
for n in nums:
    if n % 2 == 0:
        print(n)
```

# Nested Tuples

```python
students = (
    ("Ali", 80),
    ("Sara", 90),
    ("Ahmed", 75)
)

print(students[0])
print(students[0][0])
print(students[0][1])


a = (10, 20, 30)
b = list(a)

print(b)
# [10, 20, 30]
a = [10, 20, 30]
b = tuple(a)

print(b)
# (10, 20, 30)
```
