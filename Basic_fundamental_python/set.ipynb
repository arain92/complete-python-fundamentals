# Set mein value add karna — add()

```python
nums  = {20,30,30,40}
nums.add(50)
print(nums)

#Mulltiple Values in add in set

nums  = {20,30,30,40}
nums.update([90,60,100])
print(nums)



```

# Set se value remove karna

```python
nums  = {20,30,30,40}
nums.remove(20)
nums.discard(20)
print(nums)
```

#  Set mein indexing aur slicing kyun nahi hoti?


```python
numbers = [10, 20, 30]

print(numbers[0])



numbers = {10, 20, 30}

print(20 in numbers)
print(50 in numbers)
```

# Union — dono sets ki saari unique values
Union ka matlab: A aur B ki values ko combine karna.

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A&B)
print(A.intersection(B))
print(A.union(B))
```

# Subset aur Superset
Ye concepts thore advanced hain, lekin easy hain.

```python
A = {1, 2}
B = {1, 2, 3, 4}

print(A<=B)
print(A.issubset(B))

print(B.issuperset(A))
```
