# Slicing

```python
nums = [1, 2, 3, 4, 5, 6]
print(nums[1:4])
```

```python
numbers = [10, 20, 30, 40, 50, 60]
print(numbers[:3])  # Output: [10, 20, 30]
print(numbers[2:])  # Output: [30, 40, 50, 60]
print(numbers[:])   # Output: [10, 20, 30, 40, 50, 60]
print(numbers[::3])  # Output: [10, 30, 50]
print(numbers[::-1])  # Output: [60, 50, 40, 30, 20, 10]
```

```python
# reverse the list first logic
nums = [1, 2, 3, 4, 5, 6]
nums.reverse()
print(nums)  # Output: [6, 5, 4, 3, 2, 1]
# reverse the list second logic
nums = [1, 2, 3, 4, 5, 6]
reversed_nums = nums[::-1]
print(reversed_nums)  # Output: [6, 5, 4, 3, 2, 1]
```

# Appling Some Method in List More and Disscuss

```python
# add an element to the end of the list
nums = [1, 2, 3, 4, 5, 6]
nums.append(7)
print("List after appending 7:", nums)
# add an element at a specific index
nums = [1, 2, 3, 4, 5, 6]
nums.insert(2,100)
print("List after inserting 100 at index 2:", nums)
# multiple add elements to the end of the list
nums = [1, 2, 3, 4, 5, 6]
nums.extend([7,8,9,10,11])
print("List after extending with [7, 8, 9, 10, 11]:", nums)
# remove an element by value
nums = [1, 2, 3, 4, 5, 6]
nums.remove(3)
print("List after removing 3:", nums)
# remove an element by index
nums = [1, 2, 3, 4, 5, 6]
nums.pop(0)
print("List after popping element at index 0:", nums)
# INDEX VALUE DELETE
nums = [1, 2, 3, 4, 5, 6]
del nums[1]
print("List after deleting element at index 1:", nums)
# clear the entire list
nums = [1, 2, 3, 4, 5, 6]
nums.clear()
print("List after clearing:", nums)

#count the occurrences of an element
nums = [1, 2, 3, 4, 5, 6, 2, 2]
count = nums.count(2)
print("Count of 2 in the list:", count)
```

# Loop in List 

```python
num = [1, 2, 3, 4, 5, 6]
for nums in num:
    print(nums) 
```

```python
nums = [1, 2, 3, 4, 5, 6]
total = 0
for num in nums:
    total = total + num
print("Sum of all elements in the list:", total)
```

# even and Odd in sparate list

```python
nums = [1, 2, 3, 4, 5, 6]
odd = []
even = []
for num in nums:
    if num % 2 == 0:
        even.append(num)
    else:
        odd.append(num)
print("Odd numbers:", odd)
print("Even numbers:", even)        
```

# square lopic with loops 

```python
nums = [1, 2, 3, 4, 5, 6]
squared = [num ** 2 for num in nums if num % 2 == 0]
print("Squared numbers:", squared)
```
