```python
class Student:
    def __init__(self,full_name,marks):
        self.full_name = full_name
        self.marks = marks
        print("Initializing Student class")
    def hello(self):
        print("Hello from Student class", self.full_name)
    def get_marks(self):
        return self.marks
    

s1 = Student("Arham", 85)
print(s1.get_marks())
# print(s1.full_name, s1.marks)
# s1.hello()


s2 = Student("Ayaan", 90)
s1.hello()
print(s2.get_marks())
# print(s2.full_name, s2.marks)
# s2.hello()
```

```python
class Student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks

    @staticmethod
    def hello():
        print("Hello from Student class")   
    def get_avg(self):
        sum = 0
        for value in self.marks:
            sum += value
        print("Average marks:", sum/len(self.marks))           

s1 = Student("Arham", [85, 90, 95])
s1.get_avg()
s1.hello()
```

# Encapsulation

```python
class Account:
    def __init__(self,bal,acc):
        self.balance = bal
        self.account_no = acc
    def debit(self,amount):
        self.balance -= amount
        print("Rs: ", amount, "debited from account no:", self.account_no)
    def credit(self,amount):
        self.balance += amount
        print("Rs: ", amount, "credited to account no:", self.account_no)
    def get_balance(self):
        return self.balance        



acc1 = Account(1000, 123456)
del acc1
print(acc1.get_balance())
```
