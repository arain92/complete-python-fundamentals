```python
nums = [100, 4, 200, 1, 3, 2, 5, 8, 7, 6]

nums.sort()

longest = []
current = []

for n in nums:
    if current == [] or n == current[-1] + 1:
        current.append(n)
    else:
        current = [n]

    if len(current) > len(longest):
        longest = current

print(longest)
```

```python

nums = [12, 7, 9, 20, 15, 8, 6, 11]

even = []
odd = []

i = 0

while i < len(nums):
    if nums[i] % 2 == 0:
        even.append(nums[i])
    else:
        odd.append(nums[i])

    i = i + 1

print("Even numbers:", even)
print("Odd numbers:", odd)
```
