```python
nums = [100, 4, 200, 1, 3, 2, 5, 8, 7, 6]

nums.sort()

longest = []
current = []

for n in nums:
    if current == [] or n == current[-1] + 1:
        current.append(n)
    else:
        current = [n]

    if len(current) > len(longest):
        longest = current

print(longest)
```

```python

nums = [12, 7, 9, 20, 15, 8, 6, 11]

even = []
odd = []

i = 0

while i < len(nums):
    if nums[i] % 2 == 0:
        even.append(nums[i])
    else:
        odd.append(nums[i])

    i = i + 1

print("Even numbers:", even)
print("Odd numbers:", odd)
```

#  **kwargs in Python
# kwargs ka use tab hota hai jab hum function mein multiple keyword arguments   pass karna chahte hain, yani arguments ke saath names bhi dete hain.

```python
def bio(**kwargs):
    print(kwargs)

bio(name="Arham", age=23, city="Karachi")
```

Samjho kaise kaam karta hai:
- name="Arham" ek keyword argument hai.
- age=23 bhi keyword argument hai.
- **kwargs in sab ko collect karta hai.
- Function ke andar kwargs ek dictionary hoti hai

```python
def bio(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

bio(name="Arham", age=23, city="Karachi")
```

# *args in Python
# *args ka use tab hota hai jab humein function mein multiple positional arguments pass karne hon aur humein pehle se pata na ho ke kitne arguments aayenge.

```python
def add(*args):
    print(args)

add(10, 20, 30, 40)
```
