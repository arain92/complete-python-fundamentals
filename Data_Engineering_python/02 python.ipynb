```python
for temp in [1,2,3,4,5]:
    print(f'{temp} Pakistan')
```

```python
for temp in range(10): #range is a generator function that generates numbers from 0 to 9
    print(temp)
```

```python
for temp in range(5,10,1): #first included last is not included, 1 is the step size
    print(temp)
```

```python
languages = ["Python", "Java", "C++"]

for index, lang in enumerate(languages):
    print(f"Index {index}: {lang}")
```

```python
nums = [100,4,1,3,2,5,8,7,6]
nums.sort() #sorts the list in ascending order
print(nums) # longest number is at the end of the list
```

```python
nums = [100, 4, 200, 1, 3, 2, 5, 8, 7, 6]

nums.sort()

longest = []
current = []

for n in nums:
    if current == [] or n == current[-1] + 1:
        current.append(n)
    else:
        current = [n]

    if len(current) > len(longest):
        longest = current

print(longest)
```
