```python
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime,timedelta
```

```python
start_date = datetime(2024,1,1)
end_date = datetime(2024,12,31)
num_days = (end_date - start_date).days
print(start_date,num_days)
```

```python
data_list = []
for i in range(num_days):
    data_list.append(start_date.date())
    start_date += timedelta(days=1)
```

```python
data_list
```

```python
stock_price_list = (np.random.randn(num_days)).tolist
```

```python
np.random.seed(0)
stock_price_list = (np.random.randn(num_days).cumsum()*50).tolist()
```

```python
stock_price_list
```

```python
plt.Figure(figsize=(15,6))
plt.plot(data_list,stock_price_list,color="red",label="price")
plt.title("STOCK PRICE")
plt.xlabel("data")
plt.ylabel("stock Price")
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
sales = [100, 200, 150, 300, 400]

plt.plot(
    days,
    sales,
    color="red",
    marker="o",
    linestyle="--",
    linewidth=4,
    label="sales"
    
)

plt.title("Weekly Sales")
plt.xlabel("Days")
plt.ylabel("Sales")

plt.grid()
plt.legend()
plt.show()
```
