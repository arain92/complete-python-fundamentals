```python
def cal_gst(price):
    new_price = price + (price * 0.1)
    print(new_price)
cal_gst(100)    
cal_gst(300) 
```

```python
# Modularization is the process of breaking a program into smaller manageable and organized pieces. In Python, these pieces are called modules. A module is a file containing Python definitions and statements. The file name is the module name with the suffix .py added.
from math import sqrt,log2
import math
print(math.sqrt(2))
print(math.log2(16))
```

```python
# NUMBER generation in Python can be done using the random module. The random module provides functions that generate random numbers. The random module is part of the Python standard library, so you don't need to install anything to use it.


from random import random  # prints a random float between 0.0 to 1.0
print(random.randint(1, 10))
```
