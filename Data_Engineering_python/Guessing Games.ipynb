# Mini_Project: Guessing Game

```python
import random

def play_game():
    luckey_number = random.randint(1, 60)
    while True: 
        user_num  = int(input("Enter a number between 1 and 60: "))
        if user_num== luckey_number:
            print("You won!")
            break
        elif user_num < luckey_number:
            print("Your number is too low. Try again.")
        else:
            print("Your number is too high. Try again.")    

    print("Welcome to the Lucky Number Game!")

play_game()   


```
