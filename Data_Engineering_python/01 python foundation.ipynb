```python
numbers = [10, 20, 10, 30]

numbers.append(40)
print(numbers)
# [10, 20, 10, 30, 40]

numbers.insert(1, 15)
print(numbers)
# [10, 15, 20, 10, 30, 40]

numbers.remove(10)
print(numbers)
# [15, 20, 10, 30, 40]

print(numbers.count(10))
# 1

print(numbers.index(20))
# 1
```

```python
inventory = ["apple", "bread", "milk", "eggs", "cheese"]

# 1. Add Coffee to the end of the list
inventory.append("coffee")
print("After adding Coffee:", inventory)

# 2. Insert Tea at index 2
inventory.insert(2, "tea")
print("After inserting Tea:", inventory)

# 3. Remove Milk from the list
inventory.remove("milk")
print("After removing Milk:", inventory)

# 4. Sort the list alphabetically
inventory.sort()
print("Final inventory:", inventory)
```
