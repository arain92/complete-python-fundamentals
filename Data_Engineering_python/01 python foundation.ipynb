# List multiple value ek sath store karwate hai  

```python
student = ['Arham','Azam','Ali','Qasim']
student.insert(1,"Muhammad")
student.append("khan")
print(student)

for index, students in  enumerate(student):
  print(index,students)
```

# Negative indexing => -1 Its means to Count Reverse and End of the List

```python
fruits = ["Apple", "Mango", "Banana", "Orange"]

print(fruits[-1])
print(fruits[-2])
```

# Positive index	Negative index	Value
# 0	-4	Apple
# 1	-3	Mango
# 2	-2	Banana
# 3	-1	Orange
